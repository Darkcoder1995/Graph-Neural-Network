```python
# =========================
# Utilities
# =========================
def set_seed(s=42):
    torch.manual_seed(s); np.random.seed(s); random.seed(s)
```

```python
_EPOCH_S_MAX = 1_000_000_000_000        # ~1e12  (seconds threshold -> treat bigger as ms/us)
_EPOCH_MS_MAX = 1_000_000_000_000_000   # ~1e15  (milliseconds threshold -> treat bigger as us)

def _parse_epoch_like(x):
    """Return epoch seconds if x looks numeric; else None."""
    # handle pandas NA / numpy NaN / None
    if x is None or (isinstance(x, float) and math.isnan(x)) or (isinstance(x, (np.floating,)) and np.isnan(x)):
        return None
    try:
        val = float(x)
    except Exception:
        return None
    if val < 0:
        return None
    if val < _EPOCH_S_MAX:        # seconds
        return val
    if val < _EPOCH_MS_MAX:       # milliseconds
        return val / 1_000.0
    # microseconds or nanoseconds
    return val / 1_000_000.0

def _combine_split_cols(df, prefix):
    date_col = time_col = None
    lower = {c.lower(): c for c in df.columns}
    for key in [f"{prefix}_date", f"{prefix}_dt", f"{prefix}_day", f"{prefix}_ymd", f"{prefix}_ymd_str", f"{prefix}date"]:
        if key in lower: date_col = lower[key]; break
    for key in [f"{prefix}_time", f"{prefix}_hhmm", f"{prefix}_hms", f"{prefix}_hour", f"{prefix}_hhmmss", f"{prefix}time"]:
        if key in lower: time_col = lower[key]; break
    if date_col and time_col:
        return (df[date_col].astype(str).str.strip() + " " + df[time_col].astype(str).str.strip())
    return None

def _add_default_year_if_missing(s: pd.Series, default_year="2022"):
    pat = re.compile(r"^\s*\d{1,2}-\d{1,2}\s+\d{1,2}:\d{2}:\d{2}\s*$")
    needs_year = s.astype(str).str.match(pat)
    if needs_year.any():
        s = s.astype(str)
        s.loc[needs_year] = default_year + "-" + s.loc[needs_year]
    return s

def series_to_unix_seconds(df: pd.DataFrame, col: str, prefix_fallback: str):
    """
    Robustly produce Unix seconds from df[col].
    Strategy: epoch-like -> datetime (as-is) -> datetime (add year) -> split cols.
    Returns: (int64 seconds series, valid_mask)
    """
    raw = df[col]

    # 1) epoch-like (float seconds or None)
    sec = raw.map(_parse_epoch_like)
    sec = pd.Series(sec, index=df.index, dtype="float64")

    # 2) strings → datetime (UTC)
    mask = sec.isna()
    if mask.any():
        txt = raw[mask].astype(str)
        ts = pd.to_datetime(txt, errors="coerce", utc=True)   # removed infer_datetime_format
        bad = ts.isna()
        if bad.any():
            txt2 = _add_default_year_if_missing(txt.copy())
            ts2 = pd.to_datetime(txt2, errors="coerce", utc=True)
            ts[bad] = ts2[bad]
        # ns → s (float)
        sec.loc[mask] = (ts.astype("int64") / 1e9)

    # 3) split date+time fallback
    mask = sec.isna()
    if mask.any():
        combined = _combine_split_cols(df.loc[mask], prefix_fallback)
        if combined is not None:
            ts = pd.to_datetime(combined, errors="coerce", utc=True)
            sec.loc[mask] = (ts.astype("int64") / 1e9)

    valid = sec.notna() & (sec >= 0)
    return sec.fillna(-1).round().astype("int64"), valid
```

```python
def sinusoid_encoding(x: torch.Tensor, num_freqs: int = 8, max_period: float = 86400.0):
    x = x.to(torch.float32) / max_period
    f = torch.arange(num_freqs, device=x.device, dtype=x.dtype)
    ang = 2 * math.pi * (2.0 ** f.unsqueeze(0)) * x.unsqueeze(1)
    return torch.cat([torch.sin(ang), torch.cos(ang)], dim=1)
```

```python
def time_calendar_features(epoch_s: torch.Tensor):
    epoch_s = epoch_s.to(torch.long)
    hour = (epoch_s % 86400).to(torch.float32) / 3600.0
    dow  = (((epoch_s // 86400) + 4) % 7).to(torch.float32)
    return torch.stack(
        [torch.sin(2*math.pi*hour/24.0), torch.cos(2*math.pi*hour/24.0),
         torch.sin(2*math.pi*dow/7.0),   torch.cos(2*math.pi*dow/7.0)], dim=1
    )
```

```python
def bucketize(values: torch.Tensor, boundaries: List[float]) -> torch.Tensor:
    b = torch.tensor(boundaries, device=values.device, dtype=values.dtype)
    return (values.unsqueeze(1) > b.unsqueeze(0)).sum(dim=1).long()
```

```python
def pairwise_haversine(lat1, lon1, lat2, lon2):
    # lat/lon in radians; returns meters
    R = 6371000.0
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = torch.sin(dlat/2)**2 + torch.cos(lat1)*torch.cos(lat2)*(torch.sin(dlon/2)**2)
    return 2*R*torch.arcsin(torch.clamp(torch.sqrt(a), 0, 1))
```

```python
def build_fast_knn(src_xy: torch.Tensor, dst_xy: torch.Tensor, k: int = 1, batch: int = 16384):
    # src_xy: (N,2) in radians (lat,lon); dst_xy: (M,2)
    # returns indices: (N,k) nearest dst indices for each src
    N = src_xy.size(0); M = dst_xy.size(0)
    out = []
    for i in range(0, N, batch):
        sl = src_xy[i:i+batch]
        lat1 = sl[:, :1]; lon1 = sl[:, 1:2]
        lat2 = dst_xy[:, :1].T  # (1,M)
        lon2 = dst_xy[:, 1:2].T
        # broadcast to (B, M)
        d = pairwise_haversine(lat1, lon1, lat2, lon2)  # (B, M)
        idx = torch.topk(d, k=k, largest=False).indices  # (B,k)
        out.append(idx.cpu())
    return torch.vstack(out)
```

```python
# ---------- GPS column fuzzy finder (define this BEFORE LaDeLoader) ----------
_KEY_SETS = {
    "pick":  ["pickup", "accept", "start", "origin", "sender", "shop", "warehouse"],
    "drop":  ["delivery", "deliver", "dropoff", "drop", "finish", "end", "dest", "receiver", "customer"],
}
def _is_lat(name): return "lat" in name.lower()
def _is_lon(name): return any(k in name.lower() for k in ["lon","lng","long"])
def _score_side(name, side):
    n = name.lower()
    score = 0
    for k in _KEY_SETS[side]:
        if k in n: score += 2
    if _is_lat(name) or _is_lon(name): score += 1
    return score

def find_gps_columns(cols):
    """
    Returns (pick_lat, pick_lon, drop_lat, drop_lon) or (None,..) if not found.
    Heuristic: choose lat/lon with highest side-specific score.
    """
    lat_cols = [c for c in cols if _is_lat(c)]
    lon_cols = [c for c in cols if _is_lon(c)]
    if not lat_cols or not lon_cols:
        return (None, None, None, None)

    def best_pair(side):
        best = (None, None, -1)
        for la in lat_cols:
            for lo in lon_cols:
                s = _score_side(la, side) + _score_side(lo, side)
                if s > best[2]:
                    best = (la, lo, s)
        return (best[0], best[1])

    pk_lat, pk_lon = best_pair("pick")
    dp_lat, dp_lon = best_pair("drop")

    # avoid reusing same columns for both if alternatives exist
    if pk_lat == dp_lat or pk_lon == dp_lon:
        candidates = []
        for la in lat_cols:
            for lo in lon_cols:
                s = _score_side(la, "drop") + _score_side(lo, "drop")
                candidates.append((s, la, lo))
        candidates.sort(reverse=True)
        for _, la, lo in candidates:
            if la != pk_lat and lo != pk_lon:
                dp_lat, dp_lon = la, lo
                break

    if pk_lat and pk_lon and dp_lat and dp_lon:
        return (pk_lat, pk_lon, dp_lat, dp_lon)
    return (None, None, None, None)
```

```python
# =========================
# LaDe loaders
# =========================
"""
    Builds a training graph and trip tuples directly from LaDe (root-level layout).
    Prefers road-network graph. Falls back to AOI kNN graph if roads.csv schema is insufficient.
    Expected structure (any subset is fine):
      <root>/
        delivery/              # delivery_cq.csv, delivery_sh.csv, ...
        road-network/          # roads.csv
        delivery_five_cities.csv   # optional (union file)
        pickup_five_cities.csv     # optional (unused here)
        data_with_trajectory_20s/  # optional
"""
class LaDeLoader:
    """
    Works with root-level LaDe layout:
      root/
        delivery/*.csv   or  delivery_five_cities.csv
        road-network/roads.csv
        (optional) data_with_trajectory_20s/
    """
    def __init__(self, cfg):
        self.cfg = cfg
        self.root = cfg.root
        self._warnings = warnings

        self.deli_dir = os.path.join(self.root, "delivery")
        self.road_dir = os.path.join(self.root, "road-network")

        self.delivery = self._load_all_delivery()
        self.roads = self._maybe_load_roads()

    def _load_all_delivery(self) -> pd.DataFrame:
        dfs = []
        if os.path.isdir(self.deli_dir):
            for p in sorted([os.path.join(self.deli_dir, f) for f in os.listdir(self.deli_dir) if f.endswith(".csv")]):
                try:
                    dfs.append(pd.read_csv(p))
                except Exception as e:
                    print(f"[warn] Skipping {p}: {e}")

        union_path = os.path.join(self.root, "delivery_five_cities.csv")
        if os.path.exists(union_path):
            try:
                dfs.append(pd.read_csv(union_path))
            except Exception as e:
                print(f"[warn] Skipping delivery_five_cities.csv: {e}")

        if not dfs:
            raise FileNotFoundError("No delivery CSVs found under delivery/ or delivery_five_cities.csv")

        delivery = pd.concat(dfs, ignore_index=True)
        cols = set(delivery.columns)

        # timestamps (try many names)
        def pick(cands):
            for c in cands:
                if c in delivery.columns: return c
            return None
        acc_col = pick(["accept_time","accept_ts","accepted_time","order_accept_time","pickup_time","start_time"])
        del_col = pick(["delivery_time","deliver_time","delivered_time","finish_time","complete_time","drop_time","end_time"])
        if acc_col is None or del_col is None:
            # try fuzzy: any column that contains 'accept' or 'pickup' and looks time-like
            time_like = [c for c in delivery.columns if any(k in c.lower() for k in ["time","ts","timestamp","date","datetime","dt"])]
            if time_like:
                # pick first two distinct
                acc_col = acc_col or time_like[0]
                del_col = del_col or (time_like[1] if len(time_like)>1 else time_like[0])
        if acc_col is None or del_col is None:
            raise ValueError(f"Could not find accept/delivery time columns. Available: {list(delivery.columns)[:50]}")

        # robust timestamps
        accept_sec, ok_a = series_to_unix_seconds(delivery, acc_col, prefix_fallback="accept")
        deliver_sec, ok_d = series_to_unix_seconds(delivery, del_col, prefix_fallback="delivery")
        ok = ok_a & ok_d
        if (~ok).any():
            delivery = delivery.loc[ok].reset_index(drop=True)
            accept_sec = accept_sec.loc[ok].reset_index(drop=True)
            deliver_sec = deliver_sec.loc[ok].reset_index(drop=True)

        out = pd.DataFrame()
        out["accept_ts"] = accept_sec.values
        out["deliver_ts"] = deliver_sec.values
        out["eta_seconds"] = (out["deliver_ts"] - out["accept_ts"]).clip(lower=0)

        # GPS columns: first try our wide fuzzy finder
        pk_lat, pk_lon, dp_lat, dp_lon = find_gps_columns(list(delivery.columns))

        # If still None, try exact aliases used by some LaDe dumps
        def fallback_pick(names):
            for n in names:
                if n in delivery.columns: return n
            return None
        pk_lat = pk_lat or fallback_pick(["accept_lat","pickup_lat","accept_gps_lat","pickup_gps_lat","pickup_latitude"])
        pk_lon = pk_lon or fallback_pick(["accept_lng","pickup_lng","accept_gps_lng","pickup_gps_lng","pickup_longitude"])
        dp_lat = dp_lat or fallback_pick(["delivery_lat","dropoff_lat","delivery_gps_lat","drop_gps_lat","dropoff_latitude"])
        dp_lon = dp_lon or fallback_pick(["delivery_lng","dropoff_lng","delivery_gps_lng","drop_gps_lng","dropoff_longitude"])

        # assign (NaN if missing)
        for name, c in [("pick_lat", pk_lat), ("pick_lon", pk_lon), ("drop_lat", dp_lat), ("drop_lon", dp_lon)]:
            if c:
                out[name] = pd.to_numeric(delivery[c], errors="coerce")
            else:
                out[name] = np.nan

        # city if present
        city_col = None
        for cand in ["city","city_id","city_name","region","area"]:
            if cand in delivery.columns:
                city_col = cand; break
        out["city"] = delivery[city_col].astype(str) if city_col else "unknown"

        # keep sane targets
        out = out[(out["eta_seconds"] > 0) & (out["eta_seconds"] < 24*3600)].reset_index(drop=True)
        return out

    def _maybe_load_roads(self):
        path = os.path.join(self.road_dir, "roads.csv")
        if not os.path.exists(path):
            self._warnings.warn("roads.csv not found; will fall back to GPS-only graph.")
            return None
        # Robust CSV sniffer
        try:
            with open(path, "r", newline="") as f:
                sample = f.read(10000)
                dialect = csv.Sniffer().sniff(sample)
                sep = dialect.delimiter
        except Exception:
            sep = None  # let pandas guess

        # Try with sniffed sep (engine='python'); do NOT pass low_memory
        try:
            return pd.read_csv(path, sep=sep, engine="python", on_bad_lines="skip")
        except Exception:
            # Try comma, then tab
            for s in [",", "\t", ";", "|"]:
                try:
                    return pd.read_csv(path, sep=s, engine="python", on_bad_lines="skip")
                except Exception:
                    continue
            self._warnings.warn("Failed to parse roads.csv with multiple delimiters; using GPS-only graph.")
            return None

    # ----------------- Graph builders -----------------
    def build_graph_and_trips(self):
        if self._roads_usable():
            return self._build_road_graph_and_trips()
        else:
            self._warnings.warn("Using GPS-only graph (no AOI, no roads). We'll build nodes via unique GPS snaps into a kNN graph.")
            return self._build_gps_knn_graph_and_trips()

    def _roads_usable(self):
        if self.roads is None: return False
        cols = set(self.roads.columns)
        have_endpoints = any(s.issubset(cols) for s in ({"u","v"},{"src","dst"},{"from","to"},{"from_id","to_id"}))
        have_length = any(c in cols for c in ["length_m","length","edge_length","dist_m","distance"])
        return have_endpoints or have_length or len(cols) > 5  # be permissive; we can still use coords

    def _build_road_graph_and_trips(self):
        roads = self.roads
        def pick(*cands):
            for c in cands:
                if c in roads.columns: return c
            return None

        u_col = pick("u","src","from","from_id")
        v_col = pick("v","dst","to","to_id")
        len_col = pick("length_m","edge_length","length","dist_m","distance")
        spd_col = pick("speed_limit","speed","max_speed")
        hist_col= pick("hist_speed","avg_speed")

        u_lat = pick("u_lat","lat_u","u_latitude")
        u_lon = pick("u_lon","lon_u","u_longitude")
        v_lat = pick("v_lat","lat_v","v_latitude")
        v_lon = pick("v_lon","lon_v","v_longitude")

        # node ids
        uvals = roads[u_col].astype(str) if u_col else pd.Series([f"U{i}" for i in range(len(roads))])
        vvals = roads[v_col].astype(str) if v_col else pd.Series([f"V{i}" for i in range(len(roads))])
        node_ids = pd.Index(pd.concat([uvals, vvals], ignore_index=True).unique())
        id2idx = {nid: i for i, nid in enumerate(node_ids)}

        node_lat = np.zeros(len(node_ids), dtype=np.float32); node_lon = np.zeros_like(node_lat)
        node_cnt = np.zeros(len(node_ids), dtype=np.int32)

        rows = []
        for _, r in roads.iterrows():
            u = id2idx[str(r[u_col])] if u_col else 0
            v = id2idx[str(r[v_col])] if v_col else 0
            length = float(r[len_col]) if (len_col and pd.notna(r[len_col])) else 1.0
            speed = float(r[spd_col]) if (spd_col and pd.notna(r[spd_col])) else 0.0
            hist  = float(r[hist_col]) if (hist_col and pd.notna(r[hist_col])) else 0.0
            rows.append((u,v,length,speed,hist))
            if u_lat and u_lon and pd.notna(r.get(u_lat)) and pd.notna(r.get(u_lon)):
                node_lat[u]+=float(r[u_lat]); node_lon[u]+=float(r[u_lon]); node_cnt[u]+=1
            if v_lat and v_lon and pd.notna(r.get(v_lat)) and pd.notna(r.get(v_lon)):
                node_lat[v]+=float(r[v_lat]); node_lon[v]+=float(r[v_lon]); node_cnt[v]+=1

        # PyG graph
        src = torch.tensor([r[0] for r in rows], dtype=torch.long)
        dst = torch.tensor([r[1] for r in rows], dtype=torch.long)
        edge_index = torch.stack([src, dst], dim=0)

        length = torch.tensor([r[2] for r in rows], dtype=torch.float32)
        spd    = torch.tensor([r[3] for r in rows], dtype=torch.float32) if any(r[3]!=0.0 for r in rows) else None
        hist   = torch.tensor([r[4] for r in rows], dtype=torch.float32) if any(r[4]!=0.0 for r in rows) else None

        # edge features
        feats = []
        if length.numel() > 0:
            len_norm = (length - length.mean())/(length.std()+1e-6)
            feats.append(len_norm.view(-1,1))
        if spd is not None:
            feats.append(((spd - spd.mean())/(spd.std()+1e-6)).view(-1,1))
        if hist is not None:
            feats.append(((hist - hist.mean())/(hist.std()+1e-6)).view(-1,1))
        edge_cont = torch.cat(feats, dim=1) if feats else torch.zeros((len(rows),0), dtype=torch.float32)

        # length buckets (optional)
        if getattr(self.cfg, "use_buckets", True) and length.numel() > 0:
            qs = torch.quantile(length, torch.linspace(0, 1, getattr(self.cfg,"num_buckets",8)+1))[1:-1]
            bucket = (length.unsqueeze(1) > qs.unsqueeze(0)).sum(1).long().to(torch.float32).view(-1,1)
            edge_attr = torch.cat([edge_cont, bucket], dim=1)
        else:
            edge_attr = edge_cont

        # node structural feats
        x = self._node_struct_from_edge_index(len(node_ids), edge_index)

        # snap trips by GPS to nearest nodes if we have node coords
        mask = (node_cnt > 0)
        trips = self.delivery.copy()
        have_pick = trips["pick_lat"].notna().sum()>0 and trips["pick_lon"].notna().sum()>0
        have_drop = trips["drop_lat"].notna().sum()>0 and trips["drop_lon"].notna().sum()>0
        if have_pick and have_drop and mask.any():
            lat = torch.tensor(np.where(mask, node_lat, np.nan), dtype=torch.float32)
            lon = torch.tensor(np.where(mask, node_lon, np.nan), dtype=torch.float32)
            mlat = torch.nan_to_num(lat, nan=float(torch.nanmean(lat)))
            mlon = torch.nan_to_num(lon, nan=float(torch.nanmean(lon)))
            node_xy = torch.stack([mlat, mlon], dim=1) * math.pi/180.0

            def haversine_batch(a, b):
                R=6371000.0
                dlat=b[:,0]-a[:,0,None]; dlon=b[:,1]-a[:,1,None]
                aa = torch.sin(dlat/2)**2 + torch.cos(a[:,0:1])*torch.cos(b[:,0])*(torch.sin(dlon/2)**2)
                return 2*R*torch.arcsin(torch.clamp(torch.sqrt(aa),0,1))

            pk = torch.tensor(trips[["pick_lat","pick_lon"]].to_numpy(np.float32))*math.pi/180.0
            dp = torch.tensor(trips[["drop_lat","drop_lon"]].to_numpy(np.float32))*math.pi/180.0
            B=4096
            src_map=np.empty(len(trips),dtype=np.int64); dst_map=np.empty(len(trips),dtype=np.int64)
            for i in range(0,len(trips),B):
                a=pk[i:i+B]; d=haversine_batch(a,node_xy); src_map[i:i+B]=torch.argmin(d,dim=1).numpy()
                a=dp[i:i+B]; d=haversine_batch(a,node_xy); dst_map[i:i+B]=torch.argmin(d,dim=1).numpy()
            trips["src_idx"]=src_map; trips["dst_idx"]=dst_map
        else:
            # fallback: kNN GPS graph instead of AOI failure
            return self._build_gps_knn_graph_and_trips()

        data = Data(x=torch.tensor(x,dtype=torch.float32), edge_index=edge_index, edge_attr=edge_attr)
        id2idx = {i:i for i in range(data.num_nodes)}
        return data, trips, id2idx

    def _build_gps_knn_graph_and_trips(self, k=16):
        """No roads, no AOIs: build nodes as unique GPS endpoints and connect via kNN (by lat/lon)."""
        trips = self.delivery.copy()
        have_pick = trips["pick_lat"].notna().sum()>0 and trips["pick_lon"].notna().sum()>0
        have_drop = trips["drop_lat"].notna().sum()>0 and trips["drop_lon"].notna().sum()>0
        if not (have_pick and have_drop):
            raise ValueError("No usable GPS columns found for pickup/dropoff.")
        # build node set
        coords = pd.concat([
            trips[["pick_lat","pick_lon"]].rename(columns={"pick_lat":"lat","pick_lon":"lon"}),
            trips[["drop_lat","drop_lon"]].rename(columns={"drop_lat":"lat","drop_lon":"lon"})
        ], ignore_index=True).dropna()
        coords = coords.drop_duplicates().reset_index(drop=True)
        coords_rad = torch.tensor(coords.to_numpy(np.float32))*math.pi/180.0

        # map each trip endpoint to node id
        def map_points(points):
            pts = torch.tensor(points.to_numpy(np.float32))*math.pi/180.0
            B=4096; out=np.empty(len(points),dtype=np.int64)
            for i in range(0,len(points),B):
                a=pts[i:i+B]
                R=6371000.0
                dlat=coords_rad[:,0][None,:]-a[:,0:1]; dlon=coords_rad[:,1][None,:]-a[:,1:1]
                aa = torch.sin(dlat/2)**2 + torch.cos(a[:,0:1])*torch.cos(coords_rad[:,0][None,:])*(torch.sin(dlon/2)**2)
                D = 2*R*torch.arcsin(torch.clamp(torch.sqrt(aa),0,1))
                out[i:i+B]=torch.argmin(D,dim=1).numpy()
            return out

        trips["src_idx"] = map_points(trips[["pick_lat","pick_lon"]])
        trips["dst_idx"] = map_points(trips[["drop_lat","drop_lon"]])

        # kNN edges among coords
        N=len(coords); src_list=[]; dst_list=[]
        for i in range(N):
            lat1=coords_rad[i,0]; lon1=coords_rad[i,1]
            dlat=coords_rad[:,0]-lat1; dlon=coords_rad[:,1]-lon1
            R=6371000.0
            a = torch.sin(dlat/2)**2 + torch.cos(lat1)*torch.cos(coords_rad[:,0])*(torch.sin(dlon/2)**2)
            dist = 2*R*torch.arcsin(torch.clamp(torch.sqrt(a),0,1))
            idx = torch.topk(-dist, k=min(k+1,N)).indices  # largest negative -> smallest distance
            idx = [j.item() for j in idx if j!=i][:k]
            for j in idx:
                src_list.append(i); dst_list.append(j)
        edge_index = torch.tensor([src_list, dst_list], dtype=torch.long)
        # simple edge_attr = normalized distance + bucket
        dist_all=[]
        for u,v in zip(src_list,dst_list):
            dlat=coords_rad[v,0]-coords_rad[u,0]; dlon=coords_rad[v,1]-coords_rad[u,1]
            R=6371000.0
            a = torch.sin(dlat/2)**2 + torch.cos(coords_rad[u,0])*torch.cos(coords_rad[v,0])*(torch.sin(dlon/2)**2)
            dist_all.append((2*R*torch.asin(torch.clamp(torch.sqrt(a),0,1))).item())
        length=torch.tensor(dist_all,dtype=torch.float32)
        len_norm=(length-length.mean())/(length.std()+1e-6)
        qs=torch.quantile(length, torch.linspace(0,1,9))[1:-1]
        bucket=((length.unsqueeze(1)>qs.unsqueeze(0)).sum(1).float()).view(-1,1)
        edge_attr=torch.cat([len_norm.view(-1,1), bucket], dim=1)

        # node structural feats (degree/between/closeness)
        x = self._node_struct_from_edge_index(N, edge_index)
        data = Data(x=torch.tensor(x,dtype=torch.float32), edge_index=edge_index, edge_attr=edge_attr)
        id2idx={i:i for i in range(N)}
        return data, trips, id2idx

    def _node_struct_from_edge_index(self, N, edge_index):
        G = nx.DiGraph(); G.add_nodes_from(range(N))
        if edge_index.numel()>0:
            s,d = edge_index[0].cpu().numpy(), edge_index[1].cpu().numpy()
            G.add_edges_from(list(zip(s.tolist(), d.tolist())))
        deg = np.array([G.degree(i) for i in range(N)], dtype=np.float32)
        k = min(512, max(10, N//100))
        bw = nx.betweenness_centrality(G, k=k, seed=42, normalized=True) if G.number_of_edges()>0 else {}
        cl = nx.closeness_centrality(G) if G.number_of_edges()>0 else {i:0.0 for i in range(N)}
        X = np.stack([deg,
                      np.array([bw.get(i,0.0) for i in range(N)], dtype=np.float32),
                      np.array([cl.get(i,0.0) for i in range(N)], dtype=np.float32)], axis=1)
        return StandardScaler().fit_transform(X).astype(np.float32)
```

```python
# =========================
# Model (T-STGT)
# =========================
class TimeEncoder(nn.Module):
    def __init__(self, freq=8, out=64):
        super().__init__(); self.freq=freq; self.proj=nn.Linear(2*freq+4, out)
    def forward(self, depart_ts: torch.Tensor):
        return self.proj(torch.cat([sinusoid_encoding(depart_ts, self.freq),
                                    time_calendar_features(depart_ts)], dim=1))
```

```python
class EdgeAttrEncoder(nn.Module):
    def __init__(self, cont_dim: int, bucket_vocab: int, out_dim: int = 64):
        super().__init__()
        self.has_bucket = bucket_vocab > 0
        self.bucket_emb = nn.Embedding(bucket_vocab, 16) if self.has_bucket else None
        in_dim = cont_dim + (16 if self.has_bucket else 0)
        self.lin = nn.Linear(in_dim, out_dim) if in_dim>0 else nn.Identity()
    def forward(self, edge_attr: torch.Tensor):
        if edge_attr.numel()==0: return edge_attr
        cont = edge_attr[:, :-1] if self.has_bucket else edge_attr
        if self.has_bucket:
            b = edge_attr[:, -1].long().clamp(min=0)
            return self.lin(torch.cat([cont, self.bucket_emb(b)], dim=1))
        else:
            return self.lin(cont)
```

```python
class NodeInit(nn.Module):
    def __init__(self, node_in: int, time_dim: int, hidden: int):
        super().__init__(); self.fc = nn.Linear(node_in + time_dim, hidden)
    def forward(self, x: torch.Tensor, time_emb: torch.Tensor):
        t = time_emb.mean(dim=0, keepdim=True).repeat(x.size(0),1)
        return self.fc(torch.cat([x,t], dim=1))
```

```python
class TSTGTLayer(nn.Module):
    def __init__(self, hidden: int, heads: int, edge_dim: int, dropout: float):
        super().__init__()
        self.conv = TransformerConv(hidden, hidden//heads, heads=heads, edge_dim=edge_dim, beta=True, dropout=dropout)
        self.ffn  = nn.Sequential(nn.Linear(hidden, 2*hidden), nn.GELU(), nn.Dropout(dropout), nn.Linear(2*hidden, hidden))
        self.n1, self.n2 = nn.LayerNorm(hidden), nn.LayerNorm(hidden)
        self.do = nn.Dropout(dropout)
    def forward(self, x, edge_index, edge_attr):
        h = self.conv(x, edge_index, edge_attr); x = self.n1(x + self.do(h))
        h2 = self.ffn(x); return self.n2(x + self.do(h2))
```

```python
class Readout(nn.Module):
    def __init__(self, hidden: int):
        super().__init__(); self.mlp = nn.Sequential(nn.Linear(2*hidden, hidden), nn.GELU(), nn.Linear(hidden, 1))
    def forward(self, h_src, h_dst): return self.mlp(torch.cat([h_src, h_dst], dim=1)).squeeze(1)
```

```python
class TSTGT(nn.Module):
    def __init__(self, node_in: int, edge_cont_dim: int, bucket_vocab: int, cfg: CFG):
        super().__init__()
        self.time_enc = TimeEncoder(freq=cfg.time_freqs, out=64)
        self.node_init = NodeInit(node_in=node_in, time_dim=64, hidden=cfg.hidden)
        self.edge_enc  = EdgeAttrEncoder(cont_dim=edge_cont_dim, bucket_vocab=bucket_vocab, out_dim=64 if (edge_cont_dim or bucket_vocab) else 0)
        self.layers = nn.ModuleList([TSTGTLayer(cfg.hidden, cfg.heads, 64 if (edge_cont_dim or bucket_vocab) else 0, cfg.dropout)
                                     for _ in range(cfg.layers)])
        self.readout = Readout(cfg.hidden)

    def forward(self, data: Data, src_idx, dst_idx, depart_ts):
        t = self.time_enc(depart_ts)                       # (B,64)
        x = self.node_init(data.x, t)                      # (N,h)
        e = self.edge_enc(data.edge_attr) if (data.edge_attr is not None and data.edge_attr.numel()>0) else None
        for gtl in self.layers:
            x = gtl(x, data.edge_index, e)
        return self.readout(x[src_idx], x[dst_idx])
```

```python
# =========================
# Train/Eval
# =========================
def split_indices(n: int, mode: str, val_ratio: float, test_ratio: float):
    if mode == "random":
        perm = torch.randperm(n)
        n_test = int(n*test_ratio); n_val = int(n*val_ratio)
        return perm[n_test+n_val:], perm[n_test:n_test+n_val], perm[:n_test]
    else:
        idx = torch.arange(n)
        n_test = int(n*test_ratio); n_val = int(n*val_ratio)
        return idx[:-(n_test+n_val)], idx[-(n_test+n_val):-n_test], idx[-n_test:]
```

```python
def train(cfg: CFG):
    set_seed(42); os.makedirs(cfg.save_dir, exist_ok=True)
    loader = LaDeLoader(cfg)
    data, trips, _ = loader.build_graph_and_trips()
    data = data.to(cfg.device)

    # Build tensors
    src = torch.tensor(trips["src_idx"].to_numpy(), dtype=torch.long, device=cfg.device)
    dst = torch.tensor(trips["dst_idx"].to_numpy(), dtype=torch.long, device=cfg.device)
    dep = torch.tensor(trips["accept_ts"].to_numpy(), dtype=torch.long, device=cfg.device)
    y   = torch.tensor(trips["eta_seconds"].to_numpy(), dtype=torch.float32, device=cfg.device)

    # Chron split by time if requested
    if cfg.split_mode == "chron":
        trips = trips.sort_values("accept_ts").reset_index(drop=True)
        perm = torch.tensor(trips.index.to_numpy(), dtype=torch.long, device=cfg.device)
        src, dst, dep, y = src[perm], dst[perm], dep[perm], y[perm]

    tr, va, te = split_indices(len(y), cfg.split_mode, cfg.val_ratio, cfg.test_ratio)

    # Edge feature dims
    edge_attr = data.edge_attr if data.edge_attr is not None else torch.empty((0,0))
    has_bucket = (edge_attr.numel()>0 and edge_attr.size(1) >= 1)
    bucket_vocab = 0
    if has_bucket:
        # last column is bucket id (float)
        bucket_vocab = cfg.num_buckets if cfg.use_buckets else 0
    edge_cont_dim = edge_attr.size(1) - (1 if (cfg.use_buckets and edge_attr.numel()>0) else 0)
    node_in = data.x.size(1)

    model = TSTGT(node_in=node_in, edge_cont_dim=edge_cont_dim, bucket_vocab=bucket_vocab, cfg=cfg).to(cfg.device)
    opt = torch.optim.AdamW(model.parameters(), lr=cfg.lr, weight_decay=cfg.wd)

    def mae(p, t): return (p - t).abs().mean()
    def rmse(p, t): return torch.sqrt(((p - t)**2).mean())

    def run(split, idx):
        is_train = (split=="train")
        model.train(is_train)
        total_mae=total_rmse=total=0.0
        for i in range(0, idx.numel(), cfg.batch_size):
            b = idx[i:i+cfg.batch_size]
            pred = model(data, src[b], dst[b], dep[b])
            loss = mse_loss(pred, y[b])
            if is_train:
                opt.zero_grad(); loss.backward()
                nn.utils.clip_grad_norm_(model.parameters(), 2.0)
                opt.step()
            with torch.no_grad():
                total += b.numel()
                total_mae  += mae(pred, y[b]).item()*b.numel()
                total_rmse += rmse(pred, y[b]).item()*b.numel()
        return total_mae/total, total_rmse/total

    best = 1e9; patience=0
    for e in range(1, cfg.epochs+1):
        tr_mae, tr_rmse = run("train", tr)
        va_mae, va_rmse = run("val", va)
        print(f"[{e:03d}] train MAE={tr_mae:.2f}s RMSE={tr_rmse:.2f}s | val MAE={va_mae:.2f}s RMSE={va_rmse:.2f}s")
        if va_mae < best - 1e-3:
            best = va_mae; patience=0
            torch.save({"model": model.state_dict(), "cfg": vars(cfg)}, os.path.join(cfg.save_dir,"best_lade.pt"))
        else:
            patience += 1
            if patience >= 5:
                print("Early stopping."); break

    ckpt = torch.load(os.path.join(cfg.save_dir,"best_lade.pt"), map_location=cfg.device)
    model.load_state_dict(ckpt["model"])
    te_mae, te_rmse = run("test", te)
    print(f"[TEST] MAE={te_mae:.2f}s RMSE={te_rmse:.2f}s")
```

```python
if __name__ == "__main__":
    cfg = CFG(
        root="./LaDe",      # change to your LaDe path
        split_mode="chron",
        val_ratio=0.1,
        test_ratio=0.1,
        device="cuda" if torch.cuda.is_available() else "cpu",
        add_twohop=True,
        twohop_cap=32,
        use_buckets=True,
        num_buckets=8,
        hidden=128,
        heads=4,
        layers=3,
        dropout=0.1,
        lr=2e-4,
        wd=1e-4,
        epochs=10,
        batch_size=2048,
        time_freqs=8,
        save_dir="./checkpoints"
    )
    train(cfg)
```
